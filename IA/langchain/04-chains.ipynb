O Básico do LCEL

A essência do LCEL é a conexão de componentes, onde a saída de um se torna a entrada do próximo
    . Runnable: É a interface base. Prompts, LLMs e parsers são todos Runnables
    . Operador Pipe |: O principal operador para encadear Runnables, A | B significa que a saída de A é passada como entrada para B

```python
import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

chat = ChatOllama(model="llama3.2", temperature=0.7)
```

```python
prompt = ChatPromptTemplate.from_template(
    "Crie uma frase criativa sobre o animal: {animal}."
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | chat | parser
```

```python
#Executar a cadeia com uma única entrada
print("--- Saída para 'leão' ---")
print(chain.invoke({"animal": "leão"}))
```

```python
#Executar a cadeia com múltiplas entradas (batch)
#O .batch() é mais eficiente para processar várias entradas de uma vez
print("\n--- Saída para 'leão' e 'golfinho' ---")
resultados = chain.batch([{"animal": "leão"}, {"animal": "golfinho"}])
for res in resultados:
    print(res)
```

```python
#Executar em cadeia com streaming
#O .stream() retorna a resposta em pedaços, ideal para iterfaces de usuário em tempo real
print("\n--- Streaming da resposta para 'panda' ---")
for chunk in chain.stream({"animal": "panda"}):
    print(chunk, end="", flush=True)
```
