```python
from langchain_core.tools import tool
from pydantic import BaseModel, Field
```

```python
class CalcularDistanciaArgs(BaseModel):
    cidade_origem: str = Field(..., description="Cidade de origem", examples=["São Paulo", "Rio de Janeiro"])
    cidade_destino: str = Field(..., description="Cidade de destino", examples=["Porto Alegre", "Curitiba"])

@tool(args_schema=CalcularDistanciaArgs)
def calcular_distancia(cidade_origem: str, cidade_destino: str):
    """Calcula a distancia entre duas cidades"""
    return f"A distancia entre {cidade_origem} e {cidade_destino} é de 300km"
```

```python
calcular_distancia.invoke({"cidade_origem": "São Paulo",
                           "cidade_destino": "Porto Alegre"})
```
