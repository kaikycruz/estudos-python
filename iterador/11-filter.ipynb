Filter
- é usado para filtrar elementos de um iteravel com base em uma função que pode retornar T ou F
- o objetivo é permitir apenas os elementos que satisfazem um condição

```python
def e_par(x):
    return x % 2 == 0
```

```python
numeros = [1, 2, 3, 4, 5, 6]
```

```python
numeros_pares = filter(e_par, numeros)
list(numeros_pares)
```

```python
produtos = [
    {"nome": "Produto A", "Preco": 10.0, "em_estoque": True},
    {"nome": "Produto B", "Preco": 20.0, "em_estoque": False},
    {"nome": "Produto C", "Preco": 30.0, "em_estoque": True},
    {"nome": "Produto D", "Preco": 40.0, "em_estoque": False},
]
```

```python
def tem_estoque(produto):
    return produto["em_estoque"]
```

```python
produtos_disponiveis = filter(tem_estoque, produtos)
list(produtos_disponiveis)
```

```python
transacoes = [200.0, -50.0, 75.0, -30.0, 100.0, -10.0]
```

```python
def despesas(transacao):
    return transacao < 0
```

```python
registro_despesas = filter(despesas, transacoes)
list(registro_despesas)
```
