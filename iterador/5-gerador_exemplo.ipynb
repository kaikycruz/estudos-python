```python
def linhasComNumerosDePalavras(arquivo, numero):
    with open(arquivo, 'r', encoding='utf-8') as file:
        for linha in file:
            if len(linha.split()) == numero:
                yield linha.strip()
```

```python
arquivo = 'text.txt'
numeroPalavras = 2
for linha in linhasComNumerosDePalavras(arquivo, numeroPalavras):
    print(linha)
```
