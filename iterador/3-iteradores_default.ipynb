```python
import itertools
```

```python
for i in range(5):
    print(i)
```

```python
frutas = ['Maçã', 'Banana', 'Laranja', 'Morango', 'Abacaxi']
for fruta in frutas:
    print(fruta)
```

```python
pessoa = {
    'nome': 'Alice',
    'idade': 30
}

for chave in pessoa:
    print(chave, pessoa[chave])
```

```python
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
    
```

```python
nomes = ['Alice', 'Bob', 'Carlos']
idades = [30, 35, 28]

for nome, idade in zip(nomes, idades):
    print(nome, idade)
```

```python
for i in itertools.count(start=10, step=2):
    if i > 20:
        break
    print(i)
```
