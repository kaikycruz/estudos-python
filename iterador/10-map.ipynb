Map
- Função que aplica uma função a todos os itens de um iterável (uma lista por exemplo)
-É uma maneira eficiente e consisa de transformar dados

```python
def quadrado(x):
    return x ** 2
```

```python
numeros = [1, 2, 3, 4, 5]
```

```python
for n in numeros:
    print(quadrado(n))
```

```python
resultados = map(quadrado, numeros)
list(resultados)
```

```python
def celsiusParaFahrenheit(c):
    return (c * 9/5) + 32
```

```python
temperaturas_celsius = [0, 20, 37, 100]
```

```python
temperaturas_fahrenheit = map(celsiusParaFahrenheit, temperaturas_celsius)
list(temperaturas_fahrenheit)
```

```python
def padronizaNomes(nome):
    return nome.strip().lower()
```

```python
nomes = ['    Alice', 'Bob', 'Carlos', 'MARIA', 'LuCas']
```

```python
nomes_normalizados = map(padronizaNomes, nomes)
list(nomes_normalizados)
```
