Yield
- É uma palavra chave para criar geradores. Quando você usa ele em uma função, você transforma a função em um gerador, que pe um tipo especial de um iterador.
- Return: Quando é usado numa função, é encerrado e o valor é retornado. Chamadas subsequentes à função iniciam a execução novamente do código.
- Yield: Pode ser chamada novamente e retomar a execução do ponto onde foi pausada, permitindo a produção de uma sequencia de valores ao longo do tempo

```python
def numImpares(n):
    for i in range(1, n+1, 2):
        yield i
```

```python
for numero in numImpares(10):
    print(numero)
```
