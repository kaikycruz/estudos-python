```python
#gerador de expressão

quadrados = (x ** 2 for x in range(10))
for i in quadrados:
    print(i)
```

```python
#ierador infinito

def contadorInfinito():
    i = 0
    while True:
        yield i
        i += 1
contador = contadorInfinito()
for _ in range(10):
    print(next(contador))
```

```python
#coleta de valores com send()

def geradorSoma():
    total = 0
    while True:
        valor = yield total
        if valor is not None:
            total += valor
soma = geradorSoma()
next(soma)
print(soma.send(10))
print(soma.send(5))
print(soma.send(15))
```

```python
# tratamento de excessoes
def geradorExcessao():
    try:
        while True:
            valor = yield
            print(f"Valor recebido: {valor}")
    except ValueError:
        print("Valor inválido!")


g = geradorExcessao()
next(g)
g.send(10)
g.throw(ValueError)
```

```python
# cascata de geradores

def multiplicarPorDois(iterable):
    for i in iterable:
        yield i * 2

def adicionarCinco(iterable):
    for i in iterable:
        yield i + 5

numeros = range(5) # 0, 1, 2, 3, 4
resultado = adicionarCinco(multiplicarPorDois(numeros))

for r in resultado:
    print(r)
```
